We dont have labels - lets say we dont know if a text is positive, negative, neutral. 
WHAT do we do? 
We do - VADER 

calence aware dictionary for sentiment reasoning. It is sensitive to polarity (positive / negative) and intensity (strength) of emotion 

Its analysis relised on a dictionary which maos lexical features to emotion intensities called sentiment scores. 
The sentiment score of a text can be ontained by summing up the intensity of each word in the text. 

We take all the words in a doc - we calculate positive or negative emotions and sum them all up to see the final result of a document 

VADER also takes in consideration relation of a word wth its surroundng words and punctuations. 

```python
import nltk
```

```python
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
sid = SentimentIntensityAnalyzer()
```

```python
a = "This is a good movie"
```

```python
sid.polarity_scores(a)
```

This sid have 4 options - negative, positive, neutral, and compound - COMPUND IS THE OVERALL IDEA - IF SOMETHING IS POSITIVE OR NEGATIVE 
But it gives scores for all 4 options.

```python
df['scores'] = df['review'].apply(lambda review : sid.polarity_scores(review))
```

```python
df.head()
```

```python
df['compound'] = df['scores'].apply(lambda d: d['compound'])
```

```python
df.head()
```

```python
df['comp_score'] = df['compound'].apply(lambda score : 'pos' if score >= 0 else 'neg')
```

```python
df.head()
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
accuracy_score(df['label'] , df['comp_score'])
```

So - here. I checked actual label column - with my created comp_score - and why comp_score? Because we created it using sid.polarities - a new model 
So we test it - and see how it is actually working. 
This is called using ML and AI for our work - instead of creating a new model and trauining and testing it. 

```python
print(classification_report(df['label'], df['comp_score']))
```

```python
print(confusion_matrix(df['label'], df['comp_score']))
```
