# LSTMs for writting next words - 

First we need to know how to get tokens other than count vectorizer or count tokenizer or tfidf. 
We can get tokens of documents uusing Tokenizer from tensorflow.keras.preprocessing.text.Tokenizer. 

we can first fit Tokenizer to the data - text_seq in our case - which is a 25+1 - 26 words list - which starts from index 0 - goes to index 25 in first sequence. And then in second it goes from 1 index to 26. And then fron 2 to 27 and so on... creates a lot of such 26 words sequence taking words from starting and moving one index position. 


We fit this Tokenizer on list - containing list of 26 - 26 words from the whole document. 

Then we get sequences. - using texts_to_sequences of tokenizer - this gets us token numbers of words in 26 words list for each list item of text_seq. 



Once we got tokens for all the words, and we get them sequence wise too, we can now divide the test and train data, create model - in our case which was an LSTM. 

# How to make it give us new words? 
Now we got a model - that is trained on our data - with LSTM - meaning it understands context and words. Which word can come after which. 

Now, we make generate_text func. 

we get a seed text - input text. And we get num of words we need to generate. Then we generate word by word. 
Using model.predict - we get texts_to_sequences of input text, and we encode it using keras.preprocessing.sequence.pad_sequences - and then we truncate from starting in encoding. So if our text is linger than 26 words - then first words gets dropped. 

Then we predict on encoded string, and do argmax on the result to get the highest probability word. 

Then we get the word using this index and tokenizer.index_word[<index>] 

And we add this word with a space.. to our original text.
