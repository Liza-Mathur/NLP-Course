Non negative matrix factorization is an unsupervised algorithm that simultaneously performs dimensionality reduction and clustering 

We can use it in conjuction with TF-IDF to model topics across documents 



```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("L:\\Workspace\\Resources\\UPDATED_NLP_COURSE\\05-Topic-Modeling\\npr.csv" , sep = "\t")
```

```python
df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf = TfidfVectorizer(max_df = 0.95 , min_df = 2, stop_words = 'english')
```

```python
dtm = tf.fit_transform(df['Article'])
```

```python
from sklearn.decomposition import NMF
```

```python
nmf = NMF(n_components=6, random_state=42)
```

```python
nmf.fit(dtm)
```

```python
tf.get_feature_names_out()[3000]
```

```python
for index, topic in enumerate(nmf.components_):
    print(f"The top 15 words for topic numner {index} : ")
    print([tf.get_feature_names_out()[index] for index in topic.argsort()[-15:]])
    print('\n')
```

```python
topic_res = nmf.transform(dtm)
```

```python
topic_res[0].argmax()
```

```python
df['Topic'] = topic_res.argmax(axis=1)
```

```python
df.head(20)
```

```python
topic_dict = {0:'Education', 1 : 'Politics' , 2:'Health Funds', 3:'Military', 4:'Government Country',5:'Lifestyle'}
df['Topic Label'] = df['Topic'].map(topic_dict)
```
