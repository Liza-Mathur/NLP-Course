# How LDA works? 
We have a few documents. 
We decide a 'k' - number of topics - we decide it for now. 
Then we go through each document, and randomy assign each word in the document to one of the k topics. 
This gives us word distributions and topic representations. 

Then - we iterate over every word in every document, for each topic t, 
we get 
p(topic t | document d) = proportion of words in document d that are currently assigned to topic t
We get p(word w | topic t) = the proportion of assignments to topic t over all documents that come from this word w 
then we re-assign w - a new topic, where we choose t with probability p(topic t| document d) * p(word w | topic t)

This is probability that topic t generated word w
Then we have each document assigned t a topic. 
Then though it, we can find common words for each topic. "

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_df=0.9, min_df = 2, stop_words = 'english')
# it will discard words that show up in 90% of documents. That means theya re too common - and are in all the topics. 
#  min_df = word should be in at least this number of or ratio of documents
```

```python
dtm = cv.fit_transform(df['Article'])
```

```python
dtm
```

```python
from sklearn.decomposition import LatentDirichletAllocation
```

```python
lda = LatentDirichletAllocation(n_components=7, random_state=42)
```

```python
 lda.fit(dtm)
```

```python
# Grab the vocab of words
len(cv.get_feature_names_out())

#  Grab the topics 


#  Grab the highest probability words per topcs 
```
