# Sentiment Analysis. 

This is about polarity of a statement, or a document. Like tweets, reviews, etc. 
We use nltk.sentiment.vader.SentimentIntensityAnalyzer().polarity_scores(doc) to check the polarity scores.

It returns 4 scores, how much positive, negative, neutral it is, and a compund score. 
If compound is < 0 - then it is negative, if its 0 - its neutral, and if its more than 0 - then its positive.
