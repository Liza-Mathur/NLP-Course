Named entity recognition seeks to locate and classidy named entity mentions in unstructured text into pre defined categories such as person names, orghanizations, locations, medical codes, time expressions, quantities, monetary values, percentages, etc. 

Like - Our goal is to read raw text like Jim bought 300 shares of Acme Corp. in 2006. 

Here we have to add additional NER (named entity recognition) information - like 

Jim[person] bought 300 shares of Acme Corp.[organization] in 2006[Time].

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
def show_ents(doc):
    if doc.ents :
        for ent in doc.ents:
            print(ent.text + '-' + ent.label_ + '-'+ str(spacy.explain(ent.label_)))
    else :
        print("No entities found")
```

```python
doc = nlp(u"Hi how are you")
```

```python
show_ents(doc)
```

```python
doc = nlp(u"May I go to Washington, DC next May to see the Washington Monument?")
```

```python
show_ents(doc)
```

```python
doc = nlp(u"Can I please have 500 dollars of Microsoft stock?")
```

```python
show_ents(doc)
```

So this is named entity recognition. We see that okay this is maybe a person or work organization or some money or number 

```python
doc = nlp(u"Tesla to build a U.K. factory for $6 million")
```

```python
show_ents(doc)
```

Tesla it did not take as company here? Why? Because it doesnt know all the organizations. 
But worry not. We can tell them that okay take tesla as organization 'ORG'

```python
from spacy.tokens import Span
```

```python
ORG = doc.vocab.strings[u"ORG"]
```

```python
ORG
```

```python
new_entity = Span(doc, 0,1,label=ORG)
```

```python
doc.ents = list(doc.ents)+[new_entity]
```

```python
show_ents(doc)
```

See?? Now its showing Tesla in organization

We added single term. But what happens when we have to add several terms. 
Like we wanna add vacuum cleaner and vaccum-cleaner both terms. See how it adds - 

```python
doc = nlp(u"Our company created a brand new vaccum cleaner ." u"This new vaccum-cleaner is best in the show")
```

```python
show_ents(doc)
```

```python
from spacy.matcher import PhraseMatcher
```

```python
matcher = PhraseMatcher(nlp.vocab)
```

```python
phrase_list = ['vaccum cleaner','vaccum-cleaner']
phrase_patterns = [nlp(text) for text in phrase_list]
```

```python
matcher.add('newmatcher',phrase_patterns)
```

```python
found_matches = matcher(doc)
```

```python
phrase_patterns
```

```python
found_matches
```

```python
from spacy.tokens import Span
```

```python
PROD = doc.vocab.strings[u"PRODUCT"]
```

```python
found_matches
```

```python
new_ents = [Span(doc,match[1],match[2],label=PROD) for match in found_matches]
```

```python
doc.ents = list(doc.ents)+new_ents
```

```python
show_ents(doc)
```

Visualizing NER

```python
doc = nlp(u"Over the last quarter Apple sold nearly 20 Thousand iPods for a profit of $6 billion. " u"By contrast, Sony only said 8 thousand Walkman music players.")
```

```python
from spacy import displacy
```

```python
displacy.render(doc, style='ent' , jupyter=True)
```

```python
for sent in doc.sents:
    displacy.render(nlp(sent.text), style='ent',jupyter=True)
```

```python
options = {'ents':['PRODUCT','ORG','DATE']}
displacy.render(doc, style='ent',jupyter=True, options = options)
```
