What is topic modeling - 
it allows us to efficiently analyzer large volumnes of text by clustering documents into topics. 
    A large amount of text data is unlabeled, meaning we wont be apply to apply our previous supervised learning approaches ,

If we have unlabeled data, then we can attempt to discover labels. In that case of text datam this means attemptng to discover the clusters of documents, grouped together by topic


# LDA 

First we use skelarn.decomposition.LatentDirichletAllocation(n_components, random_state) to get the topic modelling 
n_components - number of subjects we select. 

### How do we use it? 
So first we get our data - and fit_transform CV on it. We get our dtm. 

Then we fit lda on dtm. 
We can get number of components using len(lda.components_)

How to get a topic? We get lda.components_[0] - this returns a topic. And if we do argsort() on this - It arranges each word in descending order by how much this particular word belongs to this topic. 

This basically groupd words. Like if government and corruption are in same group - it will give it most score for that topic. alongside others words of same topic. 

How to get results ? For each document. 
We do lda.transform(dtm). 

We get topic_res - that is for each row in the data set - I get a list - of n_components scores. That is score of how much this particular document belongs to each particular topic. We do argsort, and get the last index here- that is the highest score index, and to that component / topic - will our document belong.
