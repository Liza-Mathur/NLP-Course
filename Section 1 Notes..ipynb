# Sentence Segmentation 

How do we usually get sentences from documents? 
By using doc.sents. You use the returned list - and its usually the sentences. 
But what if I dont want to follow the sentences rule? What if I dont want to segment my sentences using full stop. I want to segment my sentences using some semi colon, or somethign else. 

I have two options. 
1. I add a segmentation rule.
   For example, I add that if mmy document's setences have ';' in it, then segment it from here, and '.' both.
   How to do that? Using @spacy.language.Language.component('set_custom_boundaries')
   we create a function here, take doc, and we check the tokens in document - and ig token.text == ';', then we set is_sent_start of token.i+1 - that is next index word start - as True.

   the full function -
   from spacy.language import Language
@Language.component("set_custom_boundaries")
def set_custom_boundaries(doc):
    for token in doc[:-1]:
        if token.text == ';':
            doc[token.i+1].is_sent_start = True
    return doc

   Then I add this in nlp pipe.
   nlp.add_pipe("set_custom_boundaries", before='parser')

   Now we have added a new custom boundary in nlp pipeline for segmnetation
