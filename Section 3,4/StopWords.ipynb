### Vocablary and Matching

```python
from spacy.matcher import Matcher
```

```python
matcher = Matcher(nlp.vocab)
```

```python
pattern1 = [{'LOWER':'solarpower'}]
pattern2 = [{'LOWER':'solar'},{'IS_PUNCT':True},{'LOWER':'power'}]
```

```python
pattern3 = [{'LOWER':'solar'},{'LOWER':'power'}]
```

This is patterns. We wanna find in some string or document to find word solar power. In diff patterns. like shown in above

```python
matcher.add('SolarPowerMatcher',[pattern1,pattern2,pattern3])
```

```python
doc = nlp(u"The Solar Power industry continues to grow as solarpower increases. Solar-power is amazing.")
```

```python
found_matches = matcher(doc)
```

```python
print(found_matches)
```

Here the first match is index 1 and 3. And the first number I get - its srting representation


Lets remove patterns from matcher

```python
matcher.remove('SolarPowerMatcher')
```

```python
pattern1 = [{'LOWER':'solarpower'}]
pattern2 = [{'LOWER':'solar'} ,{'IS_PUNCT':True, 'OP':"*"},{'LOWER':'power'}]
```

```python
matcher.add('SolarPower',[pattern1, pattern2])
```

```python
found_matches2 = matcher(doc)
```

```python
print(found_matches2)
```

```python
doc2 = nlp(u"Solar--power is solar_power")
```

```python
found_matches2 = matcher(doc2)
```
