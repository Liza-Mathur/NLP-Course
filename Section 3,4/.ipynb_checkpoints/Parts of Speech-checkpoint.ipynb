```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back")
```

```python
print(doc.text)
```

Made this a document. Now we will see POS 

```python
print(doc[4].text)
```

```python
print(doc[4].pos_)
```

```python
print(doc[4].tag_)
```

```python
for token in doc:
    print(f"{token.text:{10}} {token.pos_:{10}} {token.tag_:{10}} {spacy.explain(token.tag_)}")
```

```python
word = doc[1]
```

```python
word
```

```python
word.text
```

```python
doc = nlp(u"I read a book on NLP.")
```

```python
token = doc[1]
```

```python
print(f"{token.text:{10}} {token.pos_:{10}} {token.tag_:{10}} {spacy.explain(token.tag_)}")
```

So spacy can see the reference - context. Like I read a book - and I read books. Different context, and spacy understood it.

Count attrributes

```python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back")
```

```python
POS_counts = doc.count_by(spacy.attrs.POS)
```

```python
POS_counts
```

We can check what words we have and whats vocab on doc 

```python
doc.vocab[84].text
```

```python
doc[2].pos
```

```python
for k,v in sorted(POS_counts.items()):
    print(f"{k} {doc.vocab[k].text:{5}} {v}")
```

It says we have how many nouns puncs and all

```python
tag_counts = doc.count_by(spacy.attrs.TAG)
for k,v in sorted(tag_counts.items()):
    print(f"{k} {doc.vocab[k].text:{5}} {v}")
```

We have numerical code for fine grain tag, fine grained pos, and then freq

## Visualize POS

We use displacy

```python
doc = nlp(u"The quick brown fox jumped over the lazy dog.")
```

```python
from spacy import displacy
```

```python
displacy.render(doc,style='dep',jupyter=True)
```

```python
options = {"distance":110, 'compact':'True', 'color':'yellow','bg':'#09a3d5','font':'Times'}
```

```python
displacy.render(doc,style='dep',jupyter=True, options=options)
```

```python
doc2 = nlp(u"This is a sentence. This is another sentence, possibly longer than the other.")
```

```python
spans = list(doc2.sents)
```

```python
displacy.serve(spans, style='dep',options={'distance':110,'color':'red','bg':'green'})
```
