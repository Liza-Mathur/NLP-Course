Lets use nlp.vocab.vectors.key2row

```python
wolf = nlp.vocab['wolf'].vector
cat = nlp.vocab['cat'].vector
dog = nlp.vocab['dog'].vector
```

```python
cosine_sims = lambda x,y : 1-spatial.distance.cosine(x,y)
```

```python
new_vec = wolf - dog + cat 
```

Now I have to check similarity of this new_vec with all the vectors from nlp.vocav.vectors 

```python
x = nlp.vocab.vectors.key2row.items()
```

Now we have vectors in nlp.vocab.vectors - but we dont have word for that vector. So we get that word from key2row - so... for each key - that is hash string of word, we have a value - its vector's row number in vectors array 

```python
comp_sims = []
i = 0
for word, row in nlp.vocab.vectors.key2row.items():
    word_vec = nlp.vocab.vectors[word]
    # or word_vec = nlp.vocab.vectors.data[row]
    try:
        sim_score = cosine_sims(new_vec , word_vec)
        if sim_score > 0.2 : 
            comp_sims.append((nlp.vocab.strings[word], sim_score))
    except ValueError:
        print(word_vec.shape)
        print(nlp.vocab.strings[word])
```
