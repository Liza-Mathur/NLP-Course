Word2vec is two layer neural net that processes text. 
Inputs a text corpus and its output is a set of vectors : feature vectors for words in that corpus 

That purpose and usefulness of Word2vec is to group the vectors of similar words together in vector space. 
And so it detects similarity mathematically. 

Word2vec creates vector that are distributed numerical representations of word featuressuch as the context of individual words.
It learns naturally 

Given enough data, usage, and contexts, word2vec can make highly accurate guesses about a word's meaning based on past appearances. 

Word2vec trains words against other words that neighbor them in the input. 
1. It either does thisusing context to predict a target word - method knows as continour bag of words approach.
2. Or it ises word to predict a target context - which is called skip gram.

```python
king = nlp.vocab['King'].vector
man = nlp.vocab['man'].vector
woman = nlp.vocab['woman'].vector 
```

```python
# King - man + woman = queen or princess or highness or similar vector 
```

```python
new_vec = king - man + woman
new_vec = new_vec / np.linalg.norm(new_vec)
computed_similarities = []

for word in nlp.vocab:
    if word.has_vector :
                similarity = cosine_similarity(new_vec.reshape(1,-1) , word.vector.reshape(1,-1))[0][0]
                computed_similarities.append((word.text, similarity))
```

```python
computed_similarities = sorted(computed_similarities, key = lambda item: -item[1])
```
